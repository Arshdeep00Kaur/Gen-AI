## Text Preprocessing

Raw text data is often noisy and unstructured, containing various inconsistencies such as typos, slang, abbreviations, and irrelevant information. Preprocessing helps in:

Improving Data Quality: Removing noise and irrelevant information ensures that the data fed into the model is clean and consistent.
Enhancing Model Performance: Well-preprocessed text can lead to better feature extraction, improving the performance of NLP models.
Reducing Complexity: Simplifying the text data can reduce the computational complexity and make the models more efficient.


```python
corpus = [
    "I'm so excited about the party tonight! 🎉🥳",
    "Just finished a great workout 💪 feeling tired but happy 😊",
    "Can't believe we won the game! ⚽️🏆 What an amazing match! 👏",
    "The sunset was beautiful today 🌅 nature is amazing ❤️",
    "Working from home today with my cat 🐱 and coffee ☕",
    "This movie made me cry 😭 but the ending was perfect 👌"
]
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(corpus,columns=['text'])
```

```python
df
```

```python
df['LowerCase']=df['text'].str.lower()
```

```python
df
```

```python
# pip install emoji

# Import necessary libraries
import emoji
import string
import re

# Apply transformations in sequence
df['cleanText'] = df['LowerCase'].apply(lambda x: emoji.replace_emoji(x, ' '))                    # Remove emojis
df['cleanText'] = df['cleanText'].apply(lambda x: x.translate(str.maketrans('', '', string.punctuation)))  # Remove punctuation
df['cleanText'] = df['cleanText'].apply(lambda x: re.sub(r'\s+', ' ', x).strip())                # Remove extra whitespace

# Display cleaned DataFrame
df
```

```python
## stopwords
```

```python
import nltk
nltk.download('stopwords')
from nltk.corpus import stopwords
```

```python
stopwords=set(stopwords.words('english'))
```

```python
df['NoStopWords']=df['cleanText'].apply(lambda x: ' '.join(word for word in x.split() if word not in stopwords))
```
